```python
import pandas as pd
df = pd.read_csv("traffic_data_cleaned.csv")
df["date_time"] = pd.to_datetime(
    df["date_time"], format="mixed", dayfirst=True, errors="coerce"
)
is_weekend = df["date_time"].dt.dayofweek >= 5
df["is_holiday"] = "Working Day"
df.loc[is_weekend, "is_holiday"] = "Weekend"  
df.loc[
    (df["holiday"] != "None")
    & (df["holiday"] != "")
    & (df["holiday"].notna())
    & (df["holiday"] != "Working Day"),
    "is_holiday",
] = "National Holiday"
print(df["is_holiday"].value_counts())
df.to_csv("traffic_data_final.csv", index=False)
```
